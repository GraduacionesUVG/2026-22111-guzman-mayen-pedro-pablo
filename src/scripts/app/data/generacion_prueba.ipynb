# Generación de datos de prueba para la herramienta

**Objetivo:** construir los archivos de entrada de la aplicación ([`../app.py`](../app.py)) con los estudiantes del **conjunto de prueba**, para probar la herramienta solo con estudiantes que los modelos no vieron durante el entrenamiento.

**Entradas:** `Datos/registro_cleaned.csv`, `Datos/actividades_cleaned.csv` y `Datos/prueba.csv` · **Salidas:** `Datos/registro_prueba_herramienta.csv` (349 estudiantes) y `Datos/notas_prueba_herramienta.csv` (10,384 actividades)

```python
# Librerías
import pandas as pd
import numpy as np
```

```python
# Carga del registro, actividades y conjunto de prueba
registro = pd.read_csv('../../../Datos/registro_cleaned.csv')
notas = pd.read_csv('../../../Datos/actividades_cleaned.csv')
prueba = pd.read_csv('../../../Datos/prueba.csv')
```

Filtramos el registro y las actividades con las llaves (`Est_ID`, `Anio`) del conjunto de prueba. Usamos ambas columnas porque un repitente tiene una fila por año.

```python
# Registro y actividades solo de los estudiantes de prueba
llaves = prueba[['Est_ID', 'Anio']].drop_duplicates()

registro_herramienta = registro.merge(llaves, on=['Est_ID', 'Anio'], how='inner')
notas_herramienta = notas.merge(llaves, on=['Est_ID', 'Anio'], how='inner')
```

```python
# Dimensiones del registro de prueba
registro_herramienta.shape
```

```python
# Dimensiones de las actividades de prueba
notas_herramienta.shape
```

Guardamos los dos archivos que se suben en la aplicación.

```python
# Exportación de los archivos de entrada de la aplicación
registro_herramienta.to_csv('../../../Datos/registro_prueba_herramienta.csv')
notas_herramienta.to_csv('../../../Datos/notas_prueba_herramienta.csv')
```

Verificamos el estado real del curso de un grupo de estudiantes específicos. La lista tiene 23 `Est_ID`, pero aparecen 25 filas porque algunos tienen más de un intento.

```python
# Estudiantes a verificar
est_ids = [1854, 1527, 2327, 2332, 1560, 2462, 230, 592, 2713, 303, 1219,
                    1324, 1875, 831, 748, 135, 580, 1582, 508, 1506, 380, 584, 1298]
```

```python
# Estado real del curso de esos estudiantes
reales = registro_herramienta[registro_herramienta['Est_ID'].isin(est_ids)]
print(len(reales))  # 25 filas: algunos estudiantes tienen más de un intento
print(reales[['Est_ID', 'Anio', 'EstadoCurso']])
print(reales['EstadoCurso'].value_counts())
```
